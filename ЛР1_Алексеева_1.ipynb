### 3. Проверка пропусков

В пункте 2.3 были выявлены нулевые значения в столбце time, которые исказили результаты начального статистического анализа числового столбца speed, поэтому в первую очередь стоит обработать эти строки. 

```python
import numpy as np
df.loc[df['time'] == 0, 'time'] = np.nan
```

```python
# Проверка данных на наличие пропусков
df.isna().sum()
```

Оказывается, есть всего 4 строки с нулевым значением времени. Пропуски такого рода можно сразу удалить, поскольку в масштабе всего датасета потеря данных будет незначительна.

```python
# Удаление строк, в которых в столбцах time есть None
df = df.dropna(subset=['time'])
```

Как было замечено ранее, столбец purpose содержит много пропусков. Будем считать, что в дальнейшем исследовании данный признак не играет важной роли, поэтому пропуски можно заполнить значением по умолчанию (если полностью удалить эти строки, будет потеряна значительная часть собранных данных). Однако замена пропущенного значения одним из представленных вариантов (Meal, Meeting, Customer Visit и т.д.) сильно исказит распределение целей поездок, поэтому используется новое значение "Unknown Purpose". 

```python
df['purpose'] = df['purpose'].fillna('Unknown Purpose')
df.isna().sum()
```

После удаления строк желательно обновить индексацию, чтобы избавиться от пропусков. Метод reset_index() создаст новый датафрейм, в котором: 
- индексы исходного датафрейма станут новой колонкой с названием index ;
- все строки получат обычные индексы, уже без пропусков.

```python
df = df.reset_index(drop=True)
```

### 4. Проверка дубликатов

#### Проверка явных дубликатов

**Явными дубликатами** считаются те строки, которые полностью повторяют уже встречавшиеся в датасете. Для их поиска используется метод `duplicated()`. 
Логическая индексация - это способ фильтрации данных, при котором внутрь квадратных скобок передаётся не список имен столбцов, а последовательность значений True и False (булева маска). Она накладывается на датафрейм и возвращает только те строки, для которых в объекте series (результатом метода `duplicated()`) было True.

```python
df[df.duplicated()] # получение дубликатов с помощью логической индексации
print(df.duplicated().sum()) # подсчет количества дубликатов
```

### 5. Провека типов данных

Чтобы избавиться от неявных дубликатов в столбцах с датами (например, "01.01.2023", "2023-01-01", "Jan 1, 2023"), стоит привести значения к общему формату datetime (сейчас str).

```python
df['start_date'] = pd.to_datetime(df['start_date'], format = '%Y-%m-%d %H:%M:%S', errors='coerce')
df['end_date'] = pd.to_datetime(df['end_date'], format = '%Y-%m-%d %H:%M:%S', errors='coerce')
print(df['start_date'].dtype)
print(df['end_date'].dtype)
```

Аргумент errors='coerce' осуществляет принудительную замену значений, которые pandas не смог привести к типу datetime, на NaN (пропуск). Выясним, сколько строк "сломалось" после преобразования:

### 6. Группировка данных

Группировка данных позволяет осуществить более подробное изучение материала, чтобы затем искать связи между отдельными группами. Процесс группировки может быть выражен формулой "разделить-применитьобъединить" (split-apply-combine). Данные разбиваются на группы в соответствии с определенным признаком, затем к каждой группе применяются различные вычислительные методы, такие как подсчет количества элементов в группе с помощью метода `count()` или суммирование элементов с помощью метода `sum()` . После этого результаты объединяются в новую таблицу.
Группировка в pandas осуществляется с помощью `groupby()`:
- метод принимает в качестве аргумента название столбца, по которому нужно группировать (например, год, пол, категория).
- метод возвращает объект типа DataFrameGroupBy , то есть сгруппированные
данные, к которым можно применять различные методы (count, sum, min, max,
mean, last). Если применить к ним какой-нибудь метод pandas, они станут
новой структурой данных типа DataFrame или Series .

#### Задание 3

*` Сводная таблица (pivot_table) - среднее количество пройденных миль по каждой цели поездки (PURPOSEroute). Отсортировать по убыванию столбца MILES. Округлить значение до двух
знаков.
`*

```python
pivot_df = df.pivot_table(index='purpose', values='miles', aggfunc='mean')
pivot_df = pivot_df.sort_values(by = 'miles', ascending = False).round(2)
print(pivot_df)
```

Из полученной сводной таблицы видно, что наибольшая средняя дальность поездки зафиксирована для категории Commute (180.2 мили). Данное значение вызывает сомнения, так как Commute (поездка на работу) обычно подразумевает короткие расстояния. Причиной такого высокого среднего может быть единичный выброс в данных, ведь данная сводная таблица не учитывает частоту поездок (количество записей в каждой категории), что может привести к ошибочным выводам: категории с единичными записями могут искажать картину. Для корректной интерпретации данных необходимо дополнительно рассчитать количество поездок для каждой цели.

```python
pivot_df = df.pivot_table(
    index='purpose', 
    values='miles', 
    aggfunc=['mean', 'count']# выводится и среднее, и количество
).round(2).sort_values(by = ('mean', 'miles'), ascending = False)

print(pivot_df)
```

Лидирующее значение 180.20 миль для Commute было рассчитано на основе единичной поездки, что может указывать на ошибку или выброс данных. Среди более массовых лидируют поездки с целью визита клиента (средняя дальность 21.9 миль) и поездки на деловые встречи (15.59 миль), что выглядит реалистично.

Разделим значения длительности поездки на категории следующим образом:
- Короткие поездки:  drive_time <= 10 (меньше Q1);
- Средние поездки:  10< drive_time < 27.5 (между Q1 и Q3);
- Длинные поездки:  drive_time >= 27.5 (больше Q3);

Использование квартилей позволяет создать устойчивые к выбросам категории, отражающие реальную структуру данных: большинство поездок попадут в средний диапазон.

```python
duration_cat = ['short', 'average', 'long']

df['duration_category'] = pd.cut(df['drive_time'], bins=[0, 10, 27.5, 350], labels=duration_cat) # длина списка меток на 1 меньше кол-ва интервалов
df.head()
```

Однако долгие по продолжительности личные поездки в среднем оказались поездками на большие расстояния, чем бизнес-поездки той же категории. Воспользуемся методом qcut(), который автоматически делит набор данных на равновеликие группы. Границы категоризации проходят по квантилям.
    Это значит, что в каждый интервал попадает примерно одинаковое количество наблюдений, что полезно, когда данные распределены неравномерно или содержат выбросы.

```python
df['miles_category'], bins = pd.qcut(x=df['miles'], q=3, labels=['short', 'average', 'long'], retbins=True)
df.head()
```

```python
print(df['miles_category'].value_counts())
print(bins)
```

Распределение сбалансированно, каждая категория содержит примерно одинаковое кол-во поездок. Однако стоит учитывать, что из-за сильных выбросов (поездки на 300+ миль) категория long будет иметь очень широкий диапазон значений (например, от 9 до 310 миль).
